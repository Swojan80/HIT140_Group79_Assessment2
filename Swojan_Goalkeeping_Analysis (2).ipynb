## Statistical Interpretation

### Descriptive results

The analysis included 59 goalkeepers with recorded minutes and performance save percentage. Goalkeepers who played at least 270 minutes had a mean save percentage of 67.67% (SD = 11.31%, n = 41). Goalkeepers who played fewer than 270 minutes had a mean save percentage of 46.24% (SD = 24.74%, n = 18). The observed difference between the group means was 21.43 percentage points.

### Confidence intervals

The 95% confidence interval for the mean save percentage among goalkeepers who played at least 270 minutes was 64.10% to 71.24%. For goalkeepers who played fewer than 270 minutes, the 95% confidence interval was 33.94% to 58.55%.

The wider interval for the lower-minutes group indicates greater uncertainty, which is consistent with its smaller sample size and greater variation.

### Hypotheses

- Null hypothesis (H₀): The population mean save percentage is equal for the two playing-time groups.
- Alternative hypothesis (H₁): The population mean save percentage differs between the two playing-time groups.
- Significance level: α = 0.05.

### Welch’s two-sample t-test

Welch’s two-sample t-test produced a t-statistic of 3.5165 and a p-value of 0.002147. Because the p-value was below 0.05, the null hypothesis was rejected.

There was a statistically significant difference in mean save percentage between the two playing-time groups. Goalkeepers who played at least 270 minutes had a higher average save percentage than goalkeepers who played fewer than 270 minutes.

### Practical meaning

The results indicate that goalkeepers with greater playing time recorded higher and more consistent save percentages in this tournament dataset. However, this finding shows an association and does not prove that playing more minutes caused better performance.

### Limitations

The lower-minutes group contained only 18 usable goalkeepers and showed considerable variation. Three goalkeeper records were excluded from save-percentage analysis because their performance save percentages were missing. Save percentage may also be affected by factors not examined here, including opponent quality, defensive strength, shot difficulty and the number of shots faced.
